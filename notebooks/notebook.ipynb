```python
import csv 


with open("../data/raw/gaming_data.csv", "r", newline="", encoding="utf-8") as f:
    reader = csv.reader(f)
    header = next(reader)
    Temp = []
    for row in reader:
        Temp.append(row)
    
    Temp = sorted(Temp, key=lambda s: s[0])

with open("../data/processed/processed_gaming_data.csv", "w", newline="", encoding="utf-8") as d:
    writer = csv.writer(d)
    writer.writerow(header)
    writer.writerows(Temp)
```
